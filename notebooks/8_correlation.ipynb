### Задание 3

Для стандартного теста Пирсона и интервала на основе преобразования Фишера предполагается двумерная нормальность. Проверим распределения остаточного сахара и плотности по отдельности с помощью теста Шапиро — Уилка. Нормальность обоих распределений — необходимое, но недостаточное условие двумерной нормальности.

```python
pd.Series({
    column: stats.shapiro(df[column]).pvalue
    for column in ["residual sugar", "density"]
}, name="p-value").to_frame()
```

Вычислим коэффициент Пирсона и p-value для проверки $H_0: \rho = 0$ с помощью `stats.pearsonr`. Коэффициент можно использовать как описание линейной связи и при ненормальности; параметрические p-value и доверительный интервал в таком случае требуют осторожной интерпретации.

```python
result = stats.pearsonr(df["residual sugar"], df["density"])
f"r = {result.statistic:.4f}, p-value = {result.pvalue:.4g}"
```

Получим 95%-й доверительный интервал для коэффициента корреляции. Метод `confidence_interval` по умолчанию использует преобразование Фишера.


```python
result.confidence_interval(confidence_level=0.95)
```
